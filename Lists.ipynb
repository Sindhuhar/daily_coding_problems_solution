01.Take a list of numbers as input (space-separated) and print their sum.

Sample Input: 1 2 3 4

Expected Output: 10

```python
nums = list(map(int,input().split()))
print(sum(nums))
```

##**Explanation**

Step 1:

input() takes a line of text from the user as a string.

Step 2:

.split() breaks that string into a list of smaller strings, using spaces by default.

Step 3:

map() applies a given function to each item in a list (or other iterable).

Step 4:

Since map() returns a special object, we use list() to make it an actual list.

Step 5: sum(nums)

sum() adds up all the numbers in the list.

Step 6: print(sum(nums))

Take a list of words as input (space-separated) and print it in reverse order.

Sample Input: apple banana cherry

Expected Output: cherry banana apple

```python
words = input().split()
print(' '.join(words[::-1]))
```
